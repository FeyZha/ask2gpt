```python
def normalize_scores(values: list[float]) -> list[float]:
    total = sum(values)
    if total == 0:
        return [0.0 for _ in values]
    return [value / total for value in values]
```

```python
scores = normalize_scores([2.0, 3.0, 5.0])
print(scores)
```
